```python
import pandas as pd

# مسیر فایل اکسل را وارد کن
file_path = 'saved_as.xlsx'

# خواندن فایل اکسل
df = pd.read_excel(file_path, header=None)  # بدون سر فایل خوانده می‌شود

# استخراج اطلاعات اولین جدول (هر 29 سطر یک جدول است)
table_rows = 29
tables = [df[i:i+table_rows] for i in range(0, len(df), table_rows + 2)]

# لیست دانش آموزان با معدل >= 19
students = []

for table in tables:
    # استخراج اطلاعات جدول
    name = table.iloc[1, 2]  # C2
    father_name = table.iloc[1, 5]  # F2
    class_name = table.iloc[2, 5]  # F3
    gpa = table.iloc[25, 5]  # F26

    # بررسی شرط معدل
    if gpa >= 19:
        students.append({
            "نام و نام خانوادگی": name,
            "نام پدر": father_name,
            "نام کلاس": class_name,
            "معدل دوره": gpa
        })

# نمایش لیست دانش آموزان
for student in students:
    print(student)
```
