```python
import csv

file_path = 'saved_as.csv'

files = ["dea", "deb", "dt", "dta", "dtb", "dtj", "yea", "yeb", "yt"]

for f in files:
    # خواندن فایل CSV
    with open("data/"+ f + file_path, newline='', encoding='utf-8') as csvfile:
        reader = csv.reader(csvfile)
        rows = list(reader)

    # استخراج اطلاعات اولین جدول (هر 29 سطر یک جدول است)
    table_rows = 29
    tables = [rows[i:i+table_rows] for i in range(0, len(rows), table_rows + 2)]

    # لیست دانش آموزان با معدل >= 19
    students = []

    for table in tables:
        # استخراج اطلاعات جدول
        name = table[1][2]  # سطر 2، ستون C (index 2)
        father_name = table[1][5]  # سطر 2، ستون F (index 5)
        class_name = table[2][5]  # سطر 3، ستون F (index 5)
        gpa = float(table[25][5])  # سطر 26، ستون F (index 5)

        # بررسی شرط معدل
        if gpa >= 19:
            students.append({
                "نام و نام خانوادگی": name,
                "نام پدر": father_name,
                "نام کلاس": class_name,
                "معدل دوره": gpa
            })

    # نمایش لیست دانش آموزان
    for student in students:
        print(student)
```

```python
import csv

# لیست پوشه‌ها
folders = ['dea', 'deb', 'dt', 'dta', 'dtb', 'dtj', 'yea', 'yeb', 'yt']

# مسیر فولدر data
data_folder = 'data'

# لیست دانش آموزان با معدل >= 19
students = []

for folder in folders:
    file_path = f'{data_folder}/{folder}/saved_as.csv'

    # خواندن فایل CSV
    with open(file_path, newline='', encoding='utf-8') as csvfile:
        reader = csv.reader(csvfile)
        rows = list(reader)

    # استخراج اطلاعات اولین جدول (هر 29 سطر یک جدول است)
    table_rows = 29
    tables = [rows[i:i+table_rows] for i in range(0, len(rows), table_rows + 2)]

    for table in tables:
        # استخراج اطلاعات جدول
        name = table[1][2]  # سطر 2، ستون C (index 2)
        father_name = table[1][5]  # سطر 2، ستون F (index 5)
        class_name = table[2][5]  # سطر 3، ستون F (index 5)
        gpa = float(table[25][5])  # سطر 26، ستون F (index 5)

        # بررسی شرط معدل
        if gpa >= 19:
            students.append({
                "نام و نام خانوادگی": name,
                "نام پدر": father_name,
                "نام کلاس": class_name,
                "معدل دوره": gpa,
                "پوشه": folder
            })

# نمایش لیست دانش آموزان
for student in students:
    print(student)
```

```python
import csv

# فایل CSV خروجی
output_file = 'students_with_high_gpa.csv'

# سر فایل CSV
header = ["نام و نام خانوادگی", "نام پدر", "نام کلاس", "معدل دوره", "پوشه"]

# ذخیره داده‌ها در فایل
with open(output_file, 'w', newline='', encoding='utf-8') as csvfile:
    writer = csv.DictWriter(csvfile, fieldnames=header)
    writer.writeheader()
    writer.writerows(students)
```
